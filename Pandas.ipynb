#1. Pandas Series
###A series is a one-dimensional labeled array capable of holding any data type . The axis lables are collectively called the index .

```python
##> Crafting a series

import numpy as np
import pandas as pd

lables=['a','b','c']
my_list=[10,20,30]
arr = np.array([10,20,30])
d ={1:10, 2:20, 3:30}

print(f'Created series using list :\n{pd.Series(my_list)} \n')
print(f"Created series using numpy array :\n{pd.Series(arr)} \n")
print(f"Created series using dictionary :\n{pd.Series(d)} \n")

print(f"Labeling the series :\n{pd.Series(data=my_list, index=lables)} \n")
print(f"Labeling the series :\n{pd.Series(my_list, lables)} \n")

##? We can't create an series using 2-D numpy array because the Series itself is 1-D.
```

#2. Data Frames
###When we combine multiple series it creates a Dataframe .

####i. Creating a dataframe

```python
##> Creating Dataframe using Dictionary.
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 40],
    'City': ['New York', 'San Francisco', 'Los Angeles', 'Chicago'],
    'Salary': [50000, 60000, 70000, 80000]
}

df=pd.DataFrame(data)
print(f'DataFrame using dictionary  :\n {df}\n\n')


##> Creating DataFrams using List.
data_list = [
    ["Alice" , 25 , "New york" , 50000],
    ["Bob" , 30 , "San Francisco" , 60000],
    ["Charlie" , 35 , "Los Angeles" , 70000],
    ["David" , 40 , "Chicago" , 80000]
]

df1=pd.DataFrame(data_list, columns = ["Name" , "Age" , "City" , "Salary"])
print(f'Dataframe using list :\n{df1}\n ')
```

###ii. Selection and Indexing of columns

```python
##> Indexing in Dataframes
# For example let us take the df1 dataframe to undersatnd indexing .

print(f'DataFrame : \n{df1}\n\n')

#> If I want to target the a specific index or columns of the dataframe.Then :
print(f'Targeting the Name column : \n{df1['Name']}\n\n')

#> But if I want to acess 2 or more columns at a same time then  I have to provide the list of list .
print(f'Targeting the Name and Agr of the Dataframe : \n{df1[['Name', 'Age']]}\n')
```

#Missing Data

#i. Finding Missing data

```python
import numpy as np
import pandas as pd

missing_data={
    'A':[1,2,np.nan,4,5],
    'B':[np.nan,2,3,4,5],
    'C':[1,2,3,np.nan,np.nan],
    'D':[1,np.nan,np.nan,np.nan,5]
}

df5=pd.DataFrame(missing_data)
df5
```

```python
##> TO check wheather the value is persent in the dataframe or not we use the function 'isna()'

df5.isna()

##> It returns the value either True or Flase .
# If the value is persent then -> False
# If the value is not present then -> True
```

```python
##> And If you wnat the cout of missing vaules .
df5.isna().sum()

#> This is for the rows.
```

```python
##> But what if I want it for the coulmn , like I wnat to check that wheather there is a missing value in my any of the columns of the DataFrame .

df5.isna().any()

#> by uing any() function we we can find out that wheather there exists an empty place in the columns or not .
# If there exists -> True
# If not -> False
```

# ii. Removing Missing Values

```python
df5
```

```python
##> Using dropna()

# By using the dropna() function , the output shows a sanpshot of the removed Nan rows .
#> As in my case all the rows will be diappered as every row contanin a missing value.


df5.dropna()
```

```python
##> Passing the 'threshold' parameter to the "dropna()"
#? What does the thres do ?
# > in thershold we usually pass a number(integer) and it sreachs that
# perticular row with a non-nan value .

df5.dropna(thresh=3)


# > If we are providing the thresh more thaan the number of the columns than the
# output will be an empty DataFrame .
```

#iii. Filling missing data

```python
df5
```

```python
##> fillna() function
#> By using fillna() function we can fill a value to all of the missing palce .

## All the change that we are doing are not Temporary because I'm not passing the inplace parameter to True.

df5.fillna(0)

## now as you can see that all the missing values are now being replaced by 0.
###> Not on ly zero we can pass any value , string , integer ,float any datatype but properly.
#    Like strings must be passed in quotes only .
```

```python
###? Now what if we want to replace the value of column A to 0 and other columns with any other values ?
##> For that we have to pass the value parameter to the fillna().
#> But before that we need to make a Dictionary of values according to the rows.
#> Or we can pass the dictionary directly without creating one externally.

df5.fillna(value={'A':10 , 'B':55 , 'C':56.3 , 'D':'x'})
```

```python
##> But always filling an random value is not a soultion  .
# SO we use statstical operations like mean .
df5.fillna(df5.mean())
```

```python
df5.fillna(df5.std())
```

# Merging , Joining and Concatination

#i. Merging 2 DataFrames

```python
sample1=pd.DataFrame({
    'employee_id':[1,2,3,4,5],
    'name':['Jhon', 'Anna' , 'Peter' , 'Linda' , 'Bob'],
    'department':['HR' , 'IT' , 'Finance' , 'IT' , 'HR']
})

sample2=pd.DataFrame({
    'employee_id':[1,2,3,6,7],
    'salary':[60000 , 80000 , 65000 , 70000 , 90000],
    'bouns':[5000 , 10000 , 7000 , 8000 , 12000]
})
```

```python
sample1
```

```python
sample2
```

```python
##> In order to merge two dataframes we must have comman column in both of the DataFrame .
pd.merge(sample1,sample2)

#> Only employee_id is a comman column in both of the DaatFrame.
#> As you can see that only 3 columns appered because there are 3 comman employee_ids in bothe the DataFrames.
```

```python
#> Another parameter .
pd.merge(sample1 , sample2 , on='employee_id')
```

```python
##> there is one more parameter , and that is 'how' .
#> We can pass 4 string to "how" i.e., 'inner' , 'outer' , 'right' & 'left' .

#> Inner means to merge on the basis of comman values in the comman column.
#> Outter means to merge on the basis of non-comman values of the comman column.

pd.merge(sample1,sample2,on='employee_id',how='inner')
```

```python
pd.merge(sample1,sample2,on='employee_id',how='outer')

##? As we can see that all the cells are merged now and the column value were missing are filled with Nan( Not a number ).
```

```python
pd.merge(sample1,sample2,on='employee_id',how='left')

#> This means that we have to merge the Dataframe On the basis of the first dataframe.
```

```python
pd.merge(sample1 , sample2 , on='employee_id' , how = 'right')

#> And this means that we have to merge the dataframes on the basis of second dataframe
```

#iii. Joining 2 DataFrames

```python
##> Again we have to take two sample DataFrames.
j1=pd.DataFrame({
    'name':['Alice' , 'Bob' , 'Charlie'],
} , index=[i for i in range(1,4,1)])

j2=pd.DataFrame({
    'score':[85,90,75],
}, index=[i for i in range(2,5,1)])
```

```python
j1
```

```python
j2
```

```python
##> we can join two Pandas Series by us by the keyword ".join()".
#> Here the series before "." will accqire the first position on the DataFrame nd the series in the paranthsis will accquire the second place .

j1.join(j2)

##> Here we can see that the there is NaN in the first row it's because the join keyword concats the series serial wise and as there is no value to be placed for Alice there exists an NaN.
```

```python
##> The above cell represents the inner joining .

#> The outer one look like .

j1.join(j2,how='outer')

#> It will join the entire Series.
```

#Group by Aggergation

### i. Group by

```python
##>

data ={
    'Category':['A','B','A','B','A','B','A','B'],
    'Store':['S1','S2','S1','S1','S2','S1','S2','S2'],
    'Sales':[100,200,150,250,300,350,400,450],
    'Quantity':[5,10,15,20,25,30,35,40],
    'Date':pd.date_range('2026-01-01',periods=8)
}

df6=pd.DataFrame(data)
df6
```

```python
#? Group by category and claculate the sum of sales .
Group =df6.groupby('Category')
print(Group)
```

```python

#> As swe can see that the groupby() function has created an object .
#> Which we can't access directly , so we have to use a loop over this object .
for i , v in Group:
  print(i)
  print(v)
```
